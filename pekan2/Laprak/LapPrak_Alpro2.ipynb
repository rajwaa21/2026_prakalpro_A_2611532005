## 2.1 Dasar Teori
Dalam pemrograman, data merupakan bagian penting yang digunakan dalam membuat sebuah program. Oleh karena itu, perlu pemahaman mengenai tipe data dan variabel agar data dapat disimpan dan digunakan dengan baik sesuai kebutuhan program.

### 2.1.1. Tipe Data
Tipe data adalah sifat data yang dapat disimpan dalam bidang atau variabel tertentu. Tipe data menentukan jenis nilai apa yang diperbolehkan dan operasi apa yang dapat dilakukan pada data tersebut. Dalam bahasa pemrograman Python, terdapat berbagai tipe data yang umum digunakan, seperti numerik, string, dan boolean. Tipe data umum meliputi bilangan bulat (integer), bilangan desimal (floating-point), string untuk teks, dan boolean untuk nilai benar atau salah.

### 2.1.2 Variabel
Variabel adalah tempat yang digunakan untuk menyimpan nilai dalam sebuah program. ariabel dapat berisi berbagai jenis data, seperti angka, teks, maupun nilai benar atau salah. Nilai yang disimpan dalam variabel dapat digunakan kembali dan diolah selama program berjalan.

### 2.1.3 Numerik 
Numerik digunakan untuk menyimpan data berupa angka. Di dalam Python, numerik terdiri dari int untuk bilangan bulat, float untuk bilangan desimal, dan complex untuk bilangan kompleks.

### 2.1.4 String
merupakan tipe data yang digunakan untuk menyimpan teks atau kumpulan karakter. String dapat ditulis menggunakan tanda petik tunggal (' '), tanda petik ganda (" "), maupun tanda petik tiga (""" """) untuk teks yang terdiri dari beberapa baris (multiline).

### 2.1.5 Boolean
Boolean merepresentasikan entitas logis dan dihuni oleh dua nilai yaitu true dan false. Boolean biasanya digunakan untuk menyatakan suatu kondisi dan dapat digunakan bersama percabangan seperti if.

### 2.1.6 Input
Input adalah cara program menerima data, digunakan untuk menerima data yang dimasukkan saat program dijalankan. Pada Python, fungsi input() digunakan untuk mengambil masukan dari pengguna. Data yang diperoleh dari input() dapat diubah ke tipe data tertentu menggunakan fungsi seperti int() dan float().

### 2.1.7 Konstanta
Konstanta adalah  nilai yang tidak dapat diubah yang diketahui dan tidak dapat berubah selama masa berjalannya program. Dalam Python, konstanta ditandai menggunakan Final untuk menunjukkan bahwa suatu nilai dimaksudkan untuk tetap.

## 2.2 Kode Program

### 2.2.1 Program "Numerik_2611532005"

```python
#create a variable with integer value.
a_2005=100
print("The type of variable having value",a_2005,"is",type(a_2005))

#create a variable with float value.    
b_2005=10.2345
print("The type of variable having value",b_2005,"is",type(b_2005))

#create a variable with complex value.
c_2005=100+3j
print("The type of variable having value",c_2005,"is",type(c_2005))
```

**Penjelasan :**

1. Variabel a_2005 memberikan nilai 100.
2. print("The type of variable having value", a_2005, "is", type(a_2005)) menampilkan nilai dari variabel a_2005 sekaligus menunjukkan tipe datanya.
3. b_2005 = 10.2345. Membuat variabel b_2005 dengan nilai 10.2345.
4. print("The type of variable having value", b_2005, "is", type(b_2005)) menampilkan nilai dari b_2005 dan tipe datanya.
5. Membuat variabel c_2005 dengan nilai 100 + 3j yang termasuk tipe data complex, yaitu memiliki bagian real dan imajiner.
6. print("The type of variable having value", c_2005, "is", type(c_2005)) menampilkan nilai c_2005 dan tipe datanya.

### 2.2.3 Program "Boolean_2611532005"

```python
# deklarasi variabel dengan tipe data boolean
is_lulus_2005 = True
is_cumlaude_2005 = True

#menggunakan boolean
nilai_2005 = 85
batas_lulus_2005 = 75

# menentukan nilai boolean dari kondisi
status_kelulusan_2005 = nilai_2005 >= batas_lulus_2005 #hasilnya akan true

print("=== Check Status Kelulusan ===")
print("Nilai_2005:", nilai_2005)
print("Apakah lulus?:", status_kelulusan_2005)
if is_lulus_2005 and is_cumlaude_2005:
    print("Selamat, Anda lulus dengan predikat Cumlaude!")
```

**Penjelasan :**
1. is_lulus_2005 = True, Membuat variabel boolean dengan nilai True.
2. is_cumlaude_2005 = True, Menyimpan kondisi bahwa variabel tersebut bernilai True.
3. nilai_2005 = 85 menyimpan nilai dalam bentuk integer.
4. batas_lulus_2005 = 75 menyimpan nilai minimum yang menjadi batas kelulusan.
5. status_kelulusan_2005 = nilai_2005 >= batas_lulus_2005, Membandingkan nilai dengan batas kelulusan dan menghasilkan True atau False.
6. print("=== Check Status Kelulusan ===") menampilkan judul pada output.
7. print("Nilai_2005:", nilai_2005) menampilkan nilai yang tersimpan.
8. print("Apakah lulus?:", status_kelulusan_2005) menampilkan hasil pengecekan kelulusan.
9. if is_lulus_2005 and is_cumlaude_2005, if digunakan untuk membuat kondisi, sedangkan and berarti kedua kondisi harus bernilai True.
10. print("Selamat, Anda lulus dengan predikat Cumlaude!") menampilkan pesan jika kedua kondisi terpenuhi.

### 2.2.4 Program "Input_2611532005"

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar persegi panjang
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_2005 = input ('masukkan nilai panjang:')
lebar_2005 = input ('masukkan nilai lebar:')
luas_2005 = int(panjang_2005) * int(lebar_2005)
print("luas=", luas_2005) 
```

**Penjelasan :**
1. panjang_2005 = input('masukkan nilai panjang:'), input() berfungsi untuk memasukkan nilai panjang dari pengguna.
2. lebar_2005 = input('masukkan nilai lebar:'), input() berfungsi untuk memasukkan nilai lebar dari pengguna.
3. luas_2005 = int(panjang_2005) * int(lebar_2005), int() mengubah input menjadi bilangan bulat, kemudian panjang dikalikan dengan lebar untuk menghitung luas.
4. print("luas=", luas_2005), print() digunakan untuk menampilkan hasil luas.

### 2.2.5 Program "Konstanta_2611532005"

```python
# Buat file dengan nama konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print ("pi: %f" % (PI))
jari_2005 = float(input('Masukkan nilai jari-jari:'))
luas_2005 = PI * jari_2005 * jari_2005
print("luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2005, luas_2005))
```

**Penjelasan :**
1. from typing import Final, Mengimpor Final untuk menandai variabel sebagai konstanta.
2. PI: Final = 3.14, Membuat konstanta PI dengan nilai tetap 3.14.
3. print("pi: %f" % (PI)), Menampilkan nilai PI dengan format angka desimal.
4. jari_2005 = float(input('Masukkan nilai jari-jari:')), input() menerima nilai jari-jari, sedangkan float() mengubahnya menjadi bilangan desimal.
5. luas_2005 = PI * jari_2005 * jari_2005, Menghitung luas lingkaran menggunakan nilai PI dan jari-jari.
6. print("luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2005, luas_2005)) menampilkan jari-jari dan hasil luas dengan 2 angka di belakang koma.
