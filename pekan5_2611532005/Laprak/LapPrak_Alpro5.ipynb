Oleh karena itu, bahasa pemrograman Python menyediakan struktur kontrol perulangan (looping). Pendekatan prosedural ini memungkinkan perulangan utama (outer loop) bertindak sebagai pengendali sumbu vertikal (baris), sementara perulangan di dalamnya (inner loop) bertindak sebagai pengendali sumbu horizontal (kolom).

## 2.1 Dasar Teori
Pemahaman teoritis mengenai struktur control flow merupakan syarat krusial sebelum melakukan perancangan sintaksis dan pengujian kode. Berikut adalah penjabaran konsep-konsep fundamental yang digunakan sebagai dasar dalam memecahkan studi kasus.

### 2.1.1 Konsep Dasar Perulangan
Dalam ilmu komputer, perulangan (looping) adalah sebuah urutan instruksi komputasi yang dieksekusi secara terus-menerus dan berulang kali hingga sebuah kondisi batas tertentu terpenuhi. Penggunaan struktur perulangan memungkinkan pemrogram untuk mengotomatisasi serangkaian proses eksekusi, mengurangi duplikasi penulisan kode (code redundancy), serta meminimalkan penggunaan memori dibandingkan dengan menulis instruksi yang sama secara manual baris demi baris.

### 2.1.2 Perulangan For (For Loop)
Perulangan for dalam bahasa pemrograman Python for loop berfungsi sebagai metode iterator yang mengurutkan koleksi data berurutan (sequence), seperti list, tuple, string, atau rentang angka matematis. Blok kode di dalam perulangan for akan dieksekusi satu kali untuk setiap elemen yang terdapat di dalam koleksi data tersebut, sehingga sangat ideal untuk kondisi di mana jumlah iterasi sudah diketahui secara pasti (definite iteration).

### 2.1.3 Fungsi Range (Range Function)
Untuk merancang perulangan yang mengeksekusi blok kode dalam jumlah putaran yang spesifik, Python menggunakan fungsi bawaan range(). Fungsi ini menghasilkan deret angka berurutan secara otomatis. Secara default, urutan selalu dimulai dari angka 0, bertambah 1 pada setiap langkah (increment), dan berhenti tepat satu angka sebelum nilai batas akhir yang ditentukan (sistem zero-based indexing). 

### 2.1.4 Perulangan Bersarang (Nested Loops)
Perulangan bersarang atau nested loop adalah kondisi di mana terdapat sebuah perulangan yang diletakkan di dalam badan perulangan lainnya (perulangan dalam perulangan). Dalam mekanismenya, program akan mengeksekusi perulangan bagian dalam (inner loop) secara penuh dari awal hingga selesai untuk setiap satu putaran tunggal dari perulangan bagian luar (outer loop). Ini merupakan teknik fundamental yang digunakan ketika program perlu memproses struktur data multidimensi atau merancang koordinat spasial seperti matriks baris dan kolom.

### 2.1.5 Parameter End
Dalam bahasa pemrograman Python, fungsi bawaan print() secara otomatis akan menambahkan karakter baris baru (newline atau \n) di akhir setiap keluaran yang dicetak. Untuk memanipulasi perilaku bawaan tersebut, Python menyediakan argumen parameter kata kunci berupa end. Dengan memodifikasi nilai parameter ini, misalnya mendefinisikan end="" (string kosong) atau end=" " (spasi), pemrogram dapat mencegah kursor berpindah ke bawah, sehingga eksekusi cetakan berikutnya akan diteruskan pada baris horizontal yang sama.

### 2.2.1 Program "perulangan_for1_2611532005"

```python
# Buat file dengan nama perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2005 = int(input("Masukkan jumlah perulangan: "))

for i_2005 in range(ulang_2005):
    print(f"Perulangan ke-{i_2005}")
```

3. i_2005 digunakan sebagai penunjuk yang menyimpan nilai angka pada setiap putaran, di mana perhitungannya dimulai dari angka 0.
4. Selama perulangan berjalan, program mengeksekusi instruksi cetak untuk menampilkan pesan "Perulangan ke-" yang diikuti oleh nilai urutan saat ini dari variabel i_2005.
5. Penggabungan teks string dan nilai variabel numerik dilakukan secara langsung menggunakan format f-string tanpa perlu mengubah tipe data angkanya secara manual.
6. Program akan terus mengulangi langkah tersebut secara berurutan hingga variabel i_2005 mencapai batas akhir dari deret angka yang ditentukan, lalu program berhenti.

### 2.2.3 Program "perulangan_for3_2611532005"

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2005 = int(input("Masukkan jumlah perulangan: "))

jumlah_2005 = 0
for i_2005 in range(1, ulang_2005 + 1):
    print(i_2005, end=" ")
    jumlah_2005 = jumlah_2005 + i_2005

    if i_2005 < ulang_2005:
        print(" + ", end="")
    else:
        print(" = ", jumlah_2005, end="")
print()
print("Jumlah =", jumlah_2005)
```

**Penjelasan :**

1. Memasukkan nilai jumlah perulangan. Data yang dimasukkan diubah menjadi tipe integer menggunakan int() dan disimpan pada variabel ulang_2005.
2. Menginisialisasi jumlah_2005 dengan nilai awal 0.
3. Program memulai eksekusi perulangan for dengan fungsi range(1, ulang_2005 + 1). Variabel i_2005 akan otomatis berhitung mulai dari angka 1 hingga mencapai angka batas.
4. Program mencetak nilai i_2005 secara horizontal menggunakan end=" ". Setelah dicetak, nilai i_2005
langsung ditambahkan ke dalam jumlah_2005.
5. Program mengeksekusi percabangan if. Selama angka i_2005 masih lebih kecil dari batas akhir ulang_2005, program akan menyisipkan simbol tambah " + " di sebelah kanannya.
6. Apabila i_2005 sudah mencapai angka terakhir (kondisi else), program akan mencetak simbol " = ".
7. Setelah siklus perulangan selesai, program mencetak kesimpulan akhir berupa teks "Jumlah =" beserta nilai total di jumlah_2005.

### 2.2.4 Program "nested_for1_2611532005"

```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2005 = int(input("Masukkan nilai batas: "))
for line_2005 in range(1, batas_2005 + 1):
    for j_2005 in range(1, (-1 * line_2005 + batas_2005 + 1)):
        print(".", end="")
    print(line_2005)
```

**Penjelasan :**

2. Program mengeksekusi perulangan utama (outer loop) dengan fungsi range(1, batas_2005 + 1).
3. Di dalam setiap putaran baris, program mengeksekusi perulangan bersarang (nested loop) menggunakan variabel j_2005. Batas akhir perulangan menggunakan rumus matematika (-1 * line_2005 + batas_2005 + 1).
4. Rumus tersebut menyebabkan rentang perulangan semakin mengecil seiring bertambahnya angka baris. Selama perulangan bersarang ini berjalan, program mencetak karakter titik "." secara horizontal atau menyamping menggunakan end="".

### 2.2.5 Program "nested_for2_2611532005"

```python
# Buat file dengan nama nested_for2_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2005 = int(input("Masukkan nilai batas: "))
for i_2005 in range(1, batas_2005 + 1):
    for j_2005 in range(1, batas_2005 + 1):
        print("*", end="")
    print() # pindah ke baris berikutnya
```

### 2.2.7 Program "nested_for4_2611532005"

```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2005 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2005 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2005 = tinggi_2005
    c_2005 = a_2005
    lebar_2005 = (2 * tinggi_2005) - 2

    for i_2005 in range(1, tinggi_2005 + 1):
        b_2005 = c_2005 + 1

        for j_2005 in range(1, lebar_2005 + 1):

            # Baris atas dan bawah
            if i_2005 == 1 or i_2005 == tinggi_2005:
                if j_2005 == 1 or j_2005 == lebar_2005:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_2005 == 1 or j_2005 == lebar_2005:
                    print("|", end="")
                else:
                    if j_2005 == c_2005:
                        print("<", end="")
                    elif j_2005 == b_2005:
                        print(">", end="")
                    elif j_2005 == (lebar_2005 - c_2005):
                        print("<", end="")
                    elif j_2005 == (lebar_2005 - c_2005 + 1):
                        print(">", end="")
                    elif j_2005 > b_2005 and j_2005 < (lebar_2005 - c_2005):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli java
        a_2005 -= 2

        if a_2005 <= 0:
            c_2005 = (-a_2005 + 2)
        else:
            c_2005 = a_2005
```

**Penjelasan :**

1. Memasukkan angka tinggi pola. Program menggunakan operator modulus (% 2 != 0) untuk memvalidasi input, jika ganjil program mencetak peringatan dan berhenti, jika genap eksekusi berlanjut.
2. Program menginisialisasi variabel batas ukuran kolom horizontal pada lebar_2005, serta mengatur titik awal penempatan karakter dinamis menggunakan variabel a_2005 dan c_2005.
3. Program mengeksekusi perulangan utama i_2005 untuk mengendalikan pergerakan baris secara vertikal ke bawah, serta menetapkan nilai b_2005 sebagai batas untuk c_2005.
4. Di dalam setiap baris, program menjalankan perulangan bersarang j_2005 untuk mengevaluasi setiap titik koordinat dan mencetak karakter ke arah samping menggunakan parameter end="".
5. Percabangan pertama mengecek posisi baris. Jika merupakan baris teratas (1) atau terbawah (tinggi_2005), program mencetak bingkai horizontal pembatas berupa simbol # di sudut dan = di bagian tengah.
6. Di baris isi, program mencetak bingkai vertikal | di ujung kiri dan ujung kanan koordinat j_2005.
7. Di area dalam, program mengevaluasi deretan percabangan elif untuk membentuk dua pasang simbol < dan >. Area celah di antara sepasang simbol tersebut diisi dengan karakter titik ., sementara area luar diisi dengan spasi.
8. Setelah perulangan pencetakan kolom horizontal selesai, perintah print() dieksekusi untuk memindahkan pencetakan selanjutnya ke baris baru.
9. Di akhir setiap putaran baris, nilai a_2005 dikurangi 2. Jika nilainya mencapai titik tengah atau kurang dari 0 (<=0), rumus pada variabel c_2005 diubah untuk membalikkan arah, sehingga pola bentuk menyempit di atas kembali melebar di bagian bawah.

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

n_2005 = int(input("Masukkan ukuran skala jam pasir (N): "))

# Border Atas:
print("#", end="")
for i_2005 in range(4 * n_2005 + 5):
    print("=", end="")
print("#")

# Jam Pasir Atas
for baris_2005 in range(n_2005, 0, -1):
    print("| ", end="")
    
    for spasi_2005 in range(2 * (n_2005 - baris_2005)):
        print(" ", end="")

    for angka_2005 in range(baris_2005, 0, -1):
        print(f"{angka_2005} ", end="")
        
    print("<*>", end="")

    for angka_2005 in range(1, baris_2005 + 1):
        print(f" {angka_2005}", end="")
        
    for spasi_2005 in range(2 * (n_2005 - baris_2005)):
        print(" ", end="")
        
    print(" |")

# Poros Titik Pusat Jam Pasir 
print("|", end="")
for spasi_2005 in range(2 * n_2005 + 1):
    print(" ", end="")
    
print("<*>", end="")

for spasi_2005 in range(2 * n_2005 + 1):
    print(" ", end="")
    
print("|")

# Jam Pasir Bawah
for baris_2005 in range(1, n_2005 + 1):
    print("| ", end="")
    
    for spasi_2005 in range(2 * (n_2005 - baris_2005)):
        print(" ", end="")

    for angka_2005 in range(baris_2005, 0, -1):
        print(f"{angka_2005} ", end="")    
        
    print("<*>", end="")

    for angka_2005 in range(1, baris_2005 + 1):
        print(f" {angka_2005}", end="") 
        
    for spasi_2005 in range(2 * (n_2005 - baris_2005)):
        print(" ", end="")
        
    print(" |")

# Border Bawah
print("#", end="")
for i_2005 in range(4 * n_2005 + 5):
    print("=", end="")
print("#")
```

**Penjelasan :** 

1. Meminta masukan ukuran jam pasir menggunakan fungsi input(). Data tersebut langsung dikonversi menjadi tipe data bilangan bulat melalui fungsi int() dan disimpan ke dalam variabel n_2005.
2. Pencetakan bingkai horizontal atas dengan mencetak karakter #. Parameter end="" digunakan agar pencetakan tidak berpindah ke baris baru di bawahnya.
3. Menjalankan perulangan for dengan range(4 * n_2005 + 5). Perulangan ini mencetak karakter = secara terus-menerus ke arah samping kanan.
4. Menutup baris bingkai atas dengan mencetak karakter # yang secara otomatis memicu perpindahan kursor ke baris baru.
5. Memasuki Fase Jam Pasir Atas, program menjalankan perulangan utama (outer loop) menggunakan variabel baris_2005. Fungsi range(n_2005, 0, -1) menginstruksikan perulangan untuk bergerak mundur, yang menandakan bahwa lebar deret angka akan semakin mengecil saat dieksekusi ke bawah.
6. Di awal setiap baris pada Fase 1, program mencetak dinding pembatas vertikal kiri berupa (| ).
7. Mengeksekusi perulangan bersarang (nested loop) pertama untuk mencetak spasi pendorong di sisi kiri. Jumlah spasi dihitung melalui operasi matematis 2 * (n_2005 - baris_2005). Karena nilai baris terus mengecil, hasil selisihnya membesar, menciptakan efek spasi yang bertambah dan mendorong angka membentuk garis diagonal menyempit.
8. Mengeksekusi perulangan bersarang kedua menggunakan range(baris_2005, 0, -1) untuk mencetak deretan angka yang berhitung mundur dari batas baris saat ini hingga angka 1. Pemformatan teks f-string digunakan untuk memastikan setiap angka memiliki jarak satu spasi.
9. Mencetak karakter statis <*> tepat di poros tengah baris, berfungsi sebagai pusat palindrom pemisah antara deret angka kiri dan kanan.
10. Mengeksekusi perulangan bersarang ketiga menggunakan range(1, baris_2005 + 1) untuk mencetak deretan angka pantulan yang berhitung maju dari 1 kembali ke nilai batas baris saat ini.
11. Mengeksekusi perulangan bersarang keempat untuk mencetak spasi penyeimbang di sisi kanan menggunakan rumus matematis yang sama dengan spasi kiri. Baris ini kemudian ditutup dengan dinding vertikal kanan ( |) beserta eksekusi pindah baris.
12. Memasuki Fase Poros Titik Pusat Jam Pasir yang merupakan ekuator matriks tanpa deret angka. Program mencetak dinding kiri, diikuti perulangan spasi statis sebanyak 2 * n_2005 + 1 karakter, lalu poros pusat tunggal <*>, dilanjutkan perulangan spasi kanan dengan jumlah identik, dan ditutup oleh dinding kanan.
13. Memasuki Fase Jam Pasir Bawah, program kembali mengeksekusi perulangan utama (outer loop), namun variabel baris_2005 berhitung maju menggunakan range(1, n_2005 + 1).
14. Di dalam Fase Jam Pasir Bawah urutan eksekusi empat perulangan bersarang berjalan identik dengan Fase Jam Pasir Atas. Karena baris_2005 kini bertambah besar, rumus pengurangan spasi otomatis menghasilkan nilai yang semakin mengecil hingga habis.
15. Mengakhiri matriks dengan mencetak bingkai horizontal bawah penutup. Logika yang digunakan sama persis dengan prosedur bingkai atas, yakni iterasi pencetakan = sebanyak 4 * n_2005 + 5 yang diapit oleh karakter sudut #.

<h1 align="center" > BAB III <br> KESIMPULAN </h1>

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa penggunaan percabangan (control flow) membuat jalan sebuah program menjadi lebih dinamis, tidak sekadar lurus berurutan dari atas ke bawah. Program kini bisa merespons dan memberikan hasil yang berbeda-beda tergantung pada data yang dimasukkan oleh pengguna. Untuk membuat menu pilihan yang angkanya sudah pasti penggunaan match-case terbukti lebih rapi dan efisien daripada menulis if-elif. Sementara itu, untuk mengecek aturan yang berlapis kita bisa menggunakan nested if. Dengan cara ini, program hanya akan mengecek syarat lanjutan jika syarat pertamanya sudah terpenuhi.

Selain itu, praktikum ini juga menunjukkan bahwa kita bisa melakukan perhitungan yang ditumpuk dengan menggunakan multi-if. Tujuannya agar program mengecek semua syarat satu per satu tanpa ada syarat yang terlewat atau saling membatalkan. Agar program tidak mudah error saat pengguna salah ketik (huruf besar/kecil atau ada spasi lebih), kita bisa mengatasinya dengan fungsi .strip() dan .lower(). Hasil akhir perhitungannya pun menjadi lebih rapi seperti format uang asli berkat bantuan f-string. Secara keseluruhan, pemahaman tentang logika percabangan ini sangat penting sebagai dasar untuk membuat program otomatis yang fleksibel dan bisa diterapkan langsung di dunia nyata.
