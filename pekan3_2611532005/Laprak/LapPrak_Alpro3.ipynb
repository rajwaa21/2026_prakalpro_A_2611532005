### 1.1 Latar Belakang
Dalam bahasa pemrograman Python, manipulasi data dan alur logika program dilakukan dengan menggunakan operator. Operator adalah simbol atau kata kunci khusus yang bertugas memproses operand (nilai atau variabel). Python menyediakan berbagai klasifikasi operator untuk berbagai kebutuhan komputasi, antara lain: operator aritmatika untuk perhitungan matematis, perbandingan dan logika untuk pengambilan keputusan, assignment untuk penugasan nilai, bitwise untuk manipulasi biner, serta operator khusus(membership dan identity).

Python menyediakan operator assignment (penugasan) untuk mengalokasikan atau memperbarui nilai dalam memori dengan efisien. Untuk manipulasi data pada level paling rendah, digunakan operator bitwise yang beroperasi langsung pada bit biner. Selain itu, Python memiliki keunggulan berupa operator lainnya (khusus) seperti operator membership (in, not in) dan identity (is, is not) yang sangat adaptif digunakan untuk memeriksa keanggotaan dalam sebuah struktur data (seperti list atau string) dan identitas objek di dalam memori.

## 2.1 Dasar Teori
Dalam pemrograman, kemampuan untuk mengolah, menghitung, dan memanipulasi data merupakan bagian penting dalam membangun sebuah program yang fungsional. Oleh karena itu, diperlukan pemahaman yang baik mengenai berbagai jenis operator agar nilai yang tersimpan di dalam variabel dapat diproses dengan tepat sesuai dengan alur logika dan kebutuhan program.

### 2.1.1. Aritmatika
Operator aritmatika mengeksekusi perhitungan matematika dasar maupun lanjutan pada tipe data numerik. Operasi yang didukung mencakup penjumlahan, pengurangan, perkalian, pembagian, hingga operasi yang lebih kompleks seperti perpangkatan (eksponensial) dan sisa bagi (modulus)

### 2.1.2 Perbandingan
Operator perbandingan (relasional) digunakan untuk membandingkan dua buah operand. Evaluasi dari operator ini selalu mengembalikan nilai kebenaran berjenis boolean (True atau False), yang berfungsi sebagai penentu utama arah eksekusi program pada blok pengambilan keputusan.

### 2.1.3 Logika
Operator logika, yang terdiri dari and, or, dan not, bertugas untuk memproses dan menggabungkan beberapa ekspresi kondisional sekaligus. Operator ini mengevaluasi logika proposisional untuk menghasilkan satu nilai kebenaran akhir dari kondisi-kondisi tersebut.

### 2.1.4 Assignment
Operator assignment (penugasan) memiliki fungsi dasar untuk menetapkan suatu nilai ke dalam variabel di memori. Dalam praktiknya, Python juga memfasilitasi penugasan gabungan (augmented assignment), yakni sintaks yang menyatukan operasi aritmatika dan penugasan ke dalam satu instruksi agar penulisan kode lebih efisien.

### 2.1.5 Lainnya
Kategori ini mencakup dua fitur operator spesifik dalam Python. Pertama, operator keanggotaan (in, not in) untuk mendeteksi keberadaan suatu elemen di dalam struktur data seperti list. Kedua, operator identitas (is, is not) yang bertugas memverifikasi apakah dua variabel secara harfiah menunjuk pada alokasi objek memori yang sama.

### 2.1.6 Bitswise
Operator bitwise memanipulasi data numerik (integer) secara langsung pada tingkat representasi biner. Operasi ini tidak bekerja pada angka desimal secara utuh, melainkan mengeksekusi manipulasi logika tingkat rendah (low-level manipulation) terhadap tiap-tiap bit tunggal (0 dan 1) yang menyusun angka tersebut.

## 2.2 Kode Program

### 2.2.1 Program "Aritmatika_2611532005"

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input() 
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_2005 = int(input("Input angka-1: "))
angka2_2005 = int(input("Input angka-2: "))

# Penjumlahan
hasil_2005 = angka1_2005 + angka2_2005
print("\nOperator Penjumlahan")
print("Hasil =", hasil_2005)

# Pengurangan
hasil_2005 = angka1_2005 - angka2_2005
print("\nOperator Pengurangan")
print("Hasil =", hasil_2005)

# Perkalian
hasil_2005 = angka1_2005 * angka2_2005
print("\nOperator Perkalian")
print("Hasil =", hasil_2005)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_2005 != 0:
    hasil_2005 = angka1_2005 / angka2_2005
    print("\nOperator Pembagian")
    print("Hasil =", hasil_2005)

    hasil_2005 = angka1_2005 // angka2_2005
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_2005)

    hasil_2005 = angka1_2005 % angka2_2005
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_2005)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_2005 = angka1_2005 ** angka2_2005
print("\nOperator Pangkat")
print("Hasil =", hasil_2005)
```

**Penjelasan :**
Berikut adalah penggabungan alur logika program dengan penjelasan fungsi masing-masing operator dalam bentuk poin-poin singkat yang berurutan:

1. Memasukkan angka pertama yang kemudian diubah menjadi tipe bilangan bulat (integer) dan disimpan pada variabel angka1_2005.
2. Memasukkan angka kedua yang juga diubah menjadi tipe bilangan bulat (integer) dan disimpan pada variabel angka2_2005.
3. Menggunakan operator penjumlahan (+) yang berfungsi untuk menjumlahkan nilai dari angka pertama dan angka kedua. 
4. Menggunakan operator pengurangan (-) yang berfungsi untuk mengurangkan angka pertama dengan angka kedua. 
5. Menggunakan operator perkalian (*) yang dipakai untuk melakukan perkalian antara kedua angka tersebut. 
6. Memeriksa nilai angka kedua menggunakan struktur if-else (angka2_2005 != 0) untuk memastikan pembagi bukan angka nol, untuk mencegah error pada program.
7. Jika kondisi terpenuhi (angka kedua bukan 0), program mengeksekusi tiga operasi berturut-turut yang hasilnya bergantian disimpan pada variabel hasil_2005. 
8. Operator pembagian standar (/) yang digunakan untuk pembagian yang hasilnya selalu berupa bilangan desimal (float). 
9. Operator pembagian bulat (//) yang membagi dua angka dan membulatkan hasilnya ke bawah menjadi bilangan bulat.
10. Operator sisa bagi (%) yang berfungsi untuk mengambil nilai sisa dari operasi pembagian tersebut.
11. Jika kondisi tidak terpenuhi (angka kedua bernilai 0), program akan melompat ke blok else dan menampilkan peringatan bahwa "Angka kedua tidak boleh bernilai 0."
12. Menggunakan operator pangkat (**) yang berfungsi untuk memangkatkan angka basis (angka pertama) dengan angka eksponennya (angka kedua). Hasil akhirnya disimpan pada variabel hasil_2005 dan ditampilkan.

### 2.2.2 Program "Perbandingan_2611532005"

```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_2005 = int(input("Input angka-1: "))
angka2_2005 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_2005 = angka1_2005 > angka2_2005
print("\nOperator lebih besar dari")
print("angka1_2005 > angka2_2005", hasil_2005)

# Lebih kecil dari
hasil_2005 = angka1_2005 < angka2_2005
print("\nOperator lebih kecil dari")
print("angka1_2005 < angka2_2005", hasil_2005)

# Lebih besar dari atau sama dengan
hasil_2005 = angka1_2005 >= angka2_2005
print("\nOperator lebih besar dari atau sama dengan")
print("angka1_2005 >= angka2_2005", hasil_2005)

# Lebih kecil dari atau sama dengan
hasil_2005 = angka1_2005 <= angka2_2005
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1_2005 <= angka2_2005", hasil_2005)

# Sama dengan
hasil_2005 = angka1_2005 == angka2_2005
print("\nOperator sama dengan")
print("angka1_2005 == angka2_2005 =", hasil_2005)

# Tidak sama dengan
hasil_2005 = angka1_2005 != angka2_2005
print("\nOperator tidak sama dengan")
print("angka1_2005 != angka2_2005 =", hasil_2005)

# Tambahan: perbandingan berantai dalam Python
hasil_2005 = 0 < angka1_2005 < 100
print("\nPerbandingan berantai")
print("0 < angka1_2005 < 100 =", hasil_2005)

hasil_2005 = 0 < angka2_2005 < 100
print("0 < angka2_2005 < 100 =", hasil_2005)
```

2. Meminta pengguna memasukkan angka kedua yang juga diubah menjadi tipe bilangan bulat (*integer*) dan disimpan pada variabel angka2_2005.

4. Menggunakan operator lebih kecil dari (<) yang berfungsi untuk mengecek apakah nilai angka pertama lebih kecil dari angka kedua.     
5. Menggunakan operator lebih besar dari atau sama dengan (>=) yang berfungsi untuk mengecek apakah nilai angka pertama lebih besar atau sama nilainya dengan angka kedua.    
6. Menggunakan operator lebih kecil dari atau sama dengan (<=) yang berfungsi untuk mengecek apakah nilai angka pertama lebih kecil atau sama nilainya dengan angka kedua.    
7. Menggunakan operator sama dengan (==) yang berfungsi untuk mengecek apakah nilai angka pertama sama dengan angka kedua.

### 2.2.3 Program "Logika_2611532005"

```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator logika dalam Python

# Memasukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_2005 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_2005 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print ("\nA1_2005 =", a1_2005)
print ("A2_2005 =", a2_2005)

# Konjungsi: bernilai True jika keduanya True
hasil_2005 = a1_2005 and a2_2005
print ("\nKonjungsi (AND)")
print ("A1_2005 and A2_2005 =", hasil_2005)

# Disjungsi: bernilai True jika salah satu True
hasil_2005 = a1_2005 or a2_2005
print ("\nDisjungsi (OR)")
print ("A1_2005 or A2_2005 =", hasil_2005)

# Negasi A1: membalik nilai A1
hasil_2005 = not a1_2005
print ("\nNegasi A1_2005 (NOT)")
print ("not A1_2005 =", hasil_2005)

# Negasi A2: membalik nilai A2
hasil_2005 = not a2_2005 
print ("\nNegasi A2_2005 (NOT)")
print ("not A2_2005 =", hasil_2005)

# XOR: bernilai True jika kedua nilai berbeda
hasil_2005 = a1_2005 != a2_2005
print ("\nDisjungsi Eksklusif (XOR)")
print ("A1_2005 XOR A2_2005 =", hasil_2005)
```

**Penjelasan :**
1. Memasukkan teks (berupa "true" atau "false") untuk nilai pertama. Input tersebut dibersihkan dari spasi berlebih, strip() kemudian diubah menjadi huruf kecil lower(), lalu dievaluasi apakah sama dengan "true" untuk menghasilkan tipe data Boolean (True/False)
2. Memasukkan teks untuk nilai kedua dengan proses konversi dan evaluasi Boolean yang sama.
3. Menampilkan nilai Boolean yang telah tersimpan dari variabel a1_2005 dan a2_2005.
4. Menggunakan operator logika and yang berfungsi untuk mengevaluasi dua kondisi, dan hanya akan menghasilkan nilai True jika kedua variabel bernilai True.
5. Menggunakan operator logika or yang berfungsi untuk mengevaluasi dua kondisi, dan akan menghasilkan nilai True jika minimal salah satu variabel bernilai True.
6. Menggunakan operator logika not (Negasi) pada variabel pertama yang berfungsi untuk membalikkan nilai logika (jika awalnya True menjadi False, dan sebaliknya). 
7. Menggunakan operator logika not (Negasi) pada variabel kedua dengan fungsi membalikan nilai logika yang sama.
8. Menggunakan operator perbandingan tidak sama dengan (!=) untuk mensimulasikan logika XOR (Disjungsi Eksklusif) pada tipe data Boolean. Operasi ini berfungsi dan akan menghasilkan nilai True hanya jika kedua variabel memiliki nilai logika yang berbeda (satu True dan satu False).

### 2.2.4 Program "Assignment_2611532005"

```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_2005 = int(input("Input angka-1: "))
angka2_2005 = int(input("Input angka-2: "))

print("\nNilai awal angka1_2005 =", angka1_2005)
print("Nilai angka2_2005 =", angka2_2005)

# Assignment biasa
hasil_2005 = angka1_2005
print("\nAssignment biasa (=)")
print("Hasil =", hasil_2005)

# Assignment penambahan
hasil_2005 = angka1_2005
hasil_2005 += angka2_2005
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_2005)

# Assignment pengurangan
hasil_2005 = angka1_2005
hasil_2005 -= angka2_2005
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_2005)

# Assignment perkalian
hasil_2005 = angka1_2005
hasil_2005 *= angka2_2005
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_2005)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_2005 != 0:
    hasil_2005 = angka1_2005
    hasil_2005 /= angka2_2005
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_2005)
    # Operator tambahan
    hasil_2005 = angka1_2005
    hasil_2005 //= angka2_2005
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_2005)
    hasil_2005 = angka1_2005
    hasil_2005 %= angka2_2005
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_2005)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_2005 = angka1_2005
hasil_2005 **= angka2_2005
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_2005)
```

**Penjelasan :**
1. Memasukkan angka pertama yang kemudian diubah menjadi tipe bilangan bulat (integer) dan disimpan pada variabel angka1_2005.
2. Mememasukkan angka kedua yang juga diubah menjadi tipe bilangan bulat (integer) dan disimpan pada variabel angka2_2005.
3. Menampilkan nilai awal dari angka pertama dan angka kedua ke layar sebagai informasi bagi pengguna.
4. Menggunakan operator assignment biasa (=) yang berfungsi untuk mengisi atau menugaskan nilai dari angka1_2005 ke dalam variabel hasil_2005.
5. Menggunakan operator assignment penambahan (+=) yang berfungsi untuk menjumlahkan nilai pada hasil_2005 dengan angka2_2005, lalu langsung menyimpan hasilnya kembali ke dalam variabel hasil_2005.
6. Menggunakan operator assignment pengurangan (-=) yang berfungsi untuk mengurangkan nilai pada hasil_2005 dengan angka2_2005 dan langsung memperbarui isi variabel tersebut dengan hasil pengurangannya.
7. Menggunakan operator assignment perkalian (*=) yang berfungsi untuk mengalikan nilai pada hasil_2005 dengan angka2_2005 dan menimpa nilai variabel tersebut dengan hasil perkaliannya.
8. Memeriksa nilai angka kedua menggunakan struktur percabangan if-else (angka2_2005 != 0) untuk memastikan pembagi bukan angka nol guna mencegah terjadinya ZeroDivisionError.
9. Jika kondisi terpenuhi (angka kedua bukan 0), program mengeksekusi tiga operasi assignment berturut-turut.
10. Operator assignment pembagian (/=) untuk membagi nilai variabel dengan angka kedua dan menyimpan hasil desimalnya.
11. Operator assignment pembagian bulat (//=) untuk membagi nilai variabel dengan angka kedua dan menyimpan hasil bulatnya.
12. Operator assignment sisa bagi (%=) untuk mencari sisa bagi dan menyimpan hasilnya kembali ke variabel tersebut.
13. Jika kondisi tidak terpenuhi (angka kedua bernilai 0), program melompat ke blok else dan menampilkan peringatan bahwa pembagian tidak dapat dilakukan.
14. Menggunakan operator assignment perpangkatan (**=) yang berfungsi untuk memangkatkan nilai hasil_2005 dengan angka kedua, lalu memperbarui isi variabel tersebut dengan hasil pangkatnya.

### 2.2.5 Program "Lainnya_2611532005"

```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("========================================")
print("1. OPERATOR KEANGGOTAAN")
print("========================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_2005 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_2005 = [int(angka.strip()) for angka in input_data_2005.split(",")]

nilai_dicari_2005 = int(input("Masukkan nilai yang ingin dicari: "))    

# Operator in
hasil_2005 = nilai_dicari_2005 in data_2005
print("\nOperator keanggotaan (in)")
print(nilai_dicari_2005, "in data =", hasil_2005)

# Operator not in
hasil_2005 = nilai_dicari_2005 not in data_2005
print("\nOperator keanggotaan (not in)")
print(nilai_dicari_2005, "not in data =", hasil_2005)


print("\n========================================")
print("2. OPERATOR IDENTITAS")  
print("========================================")

# objek1 menggunakan list dari input pengguna
objek1_2005 = data_2005

# objek2 menggunakan list yang sama dengan objek1
objek2_2005 = objek1_2005

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_2005 = data_2005.copy()

print("objek1_2005 =", objek1_2005)
print("objek2_2005 =", objek2_2005)
print("objek3_2005 =", objek3_2005)

# Operator is
hasil_2005 = objek1_2005 is objek2_2005
print("\nOperator identitas IS")
print("objek1_2005 is objek2_2005 =", hasil_2005)

# Operator is not
hasil_2005 = objek1_2005 is not objek3_2005
print("\nOperator identitas IS NOT")
print("objek1_2005 is not objek3_2005 =", hasil_2005)

# Membandingkan identitas dan nilai
print("\nMembandingkan identitas dan nilai")
print("objek1_2005 is objek3_2005:", objek1_2005 is objek3_2005)
print("objek1_2005 == objek3_2005:", objek1_2005 == objek3_2005)
```

**Penjelasan :**

### 2.2.6 Program "Bitwise_2611532005"

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("\n=================================")
print("3. OPERATOR BITWISE")
print("=================================") 

angka1_2005 = int(input("Masukkan angka bitwise-1"))
angka2_2005 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_2005, "| biner =", bin(angka1_2005))
print("angka2 =", angka2_2005, "| biner =", bin(angka2_2005))

# Bitwise AND
hasil_2005 = angka1_2005 & angka2_2005
print("\nBitwise AND (&)")
print(angka1_2005, "&", angka2_2005, "=", hasil_2005)
print("Biner hasil =", bin(hasil_2005))
print("Biner hasil (8 bit) =", format(hasil_2005, "08b"))

# Bitwise OR
hasil_2005 = angka1_2005 | angka2_2005
print("\nBitwise OR (|)")
print(angka1_2005, "|", angka2_2005, "=", hasil_2005)
print("Biner hasil =", bin(hasil_2005))
print("Biner hasil (8 bit) =", format(hasil_2005, "08b"))

# Bitwise XOR
hasil_2005 = angka1_2005 ^ angka2_2005
print("\nBitwise XOR (^)")
print(angka1_2005, "^", angka2_2005, "=", hasil_2005)
print("Biner hasil =", bin(hasil_2005))
print("Biner hasil (8 bit) =", format(hasil_2005, "08b"))

# Bitwise NOT
hasil_2005 = ~angka1_2005
print("\nBitwise NOT (~)")
print("~", angka1_2005, "=", hasil_2005)
print("Biner hasil =", bin(hasil_2005))
print("Biner hasil (8 bit) =", format(hasil_2005, "08b"))

# Bitwise geser kiri
jumlah_geser_2005 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2005 = angka1_2005 << jumlah_geser_2005
print("\nBitwise geser kiri (<<)")
print(angka1_2005, "<<", jumlah_geser_2005, "=", hasil_2005)
print("Biner hasil =", bin(hasil_2005))
print("Biner hasil (8 bit) =", format(hasil_2005, "08b"))

# Bitwise geser kanan
hasil_2005 = angka1_2005 >> jumlah_geser_2005
print("\nBitwise geser kanan (>>)")
print(angka1_2005, ">>", jumlah_geser_2005, "=", hasil_2005)
print("Biner hasil =", bin(hasil_2005))
print("Biner hasil (8 bit) =", format(hasil_2005, "08b"))
```

**Penjelasan :**
1. Meminta input dua buah angka (integer) dari pengguna, lalu menampilkannya dalam format desimal dan biner sebagai data awal.
2. Menggunakan operator AND (&): Menghasilkan bit 1 hanya jika kedua bit yang dibandingkan sama-sama bernilai 1.
3. Menggunakan operator OR (|): Menghasilkan bit 1 jika minimal salah satu bit bernilai 1.
4. Menggunakan operator XOR (^): Menghasilkan bit 1 hanya jika kedua bit yang dibandingkan memiliki nilai yang berbeda.
5. Menggunakan operator NOT (~): Membalikkan (inversi) seluruh nilai bit pada angka pertama (0 menjadi 1, dan sebaliknya).
6. Meminta input dari pengguna untuk menentukan jumlah pergeseran bit.
7. Menggunakan operator Geser Kiri / Left Shift (<<): Menggeser susunan bit ke kiri. (Secara matematis, setiap pergeseran setara dengan mengalikan angka tersebut dengan 2).
8. Menggunakan operator Geser Kanan / Right Shift (>>): Menggeser susunan bit ke kanan. (Secara matematis, setiap pergeseran setara dengan membagi bulat angka tersebut dengan 2).

```python
print ("\n=== SISTEM TRANSAKSI TOKO ===")

nama_2005 = input("Masukkan Nama Pelanggan: ")
status_2005 = input("Masukkan Status Pelanggan (member/nonmember): ")
total_belanja_2005 = int(input("Masukkan Total Belanja : "))
jumlah_barang_2005 = int(input("Masukkan Jumlah Barang : "))
promo_2005 = input("Masukkan Kode Promo : ")

print("\n=== DATA TRANSAKSI ===")
print("Nama Pelanggan                     : ",nama_2005)
print("Status Pelanggan (member/nonmember): ", status_2005)
print(f"Total Belanja                      :  Rp{total_belanja_2005}")
print("Jumlah Barang                      : ", jumlah_barang_2005)
print("Kode Promo                         : ", promo_2005)

kode_promo_2005 = ["HEMAT10", "HEMAT20", "GRATISONGKIR"]

syarat_total_2005 = total_belanja_2005 >=200000
syarat_jumlah_barang_2005 = jumlah_barang_2005 >=3
valid_status_2005 = status_2005 == "member"
valid_promo_2005 = promo_2005 in kode_promo_2005 

print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000                : {syarat_total_2005}") 
print(f"Jumlah Barang >= 3                 : {syarat_jumlah_barang_2005}")
print(f"Status Member                      : {valid_status_2005}")
print(f"Kode Promo Tersedia                : {valid_promo_2005}")
print(f"Mendapatkan Diskon                 : {syarat_jumlah_barang_2005 or syarat_total_2005}")
print(f"Mendapatkan Promo                  : {valid_promo_2005}")

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                             : Rp{int(0.10 * total_belanja_2005)}")
print(f"Total Pembayaran                   : Rp{int(total_belanja_2005 - 0.10 * total_belanja_2005)}")
print(f"Rata-rata Harga Barang             : Rp{int((total_belanja_2005 - 0.10 * total_belanja_2005) // jumlah_barang_2005)}")

print("\n--- HAK AKSES PELANGGAN ---")
print("Kode Hak Akses         : ...")
print(f"Member Access          : {status_2005 == 'member'}")
print(f"Promo Access           : {promo_2005 in kode_promo_2005}")
print("Free Shipping Access   : ...")

print("\n=== OPERASI BITWISE ===")

print("\n=== Kode Status Transaksi ===")

# Kondisi
# 0001 = Pelanggan merupakan member
# 0010 = Total belanja >= Rp 200000
# 0100 = Jumlah barang >= 3
# 1000 = Kode promo tersedia

kode_transaksi_2005 = int(valid_status_2005) << 0 | int(syarat_total_2005) << 1 | int(syarat_jumlah_barang_2005) << 2 | int(valid_promo_2005) << 3
kode_referensi_2005 = int(valid_status_2005) << 0 | int(syarat_total_2005) << 1 | int(valid_promo_2005) << 3

print(f"{format(int(valid_status_2005) << 0, '04b')} | {format(int(syarat_total_2005) << 1, '04b')} | {format(int(syarat_jumlah_barang_2005) << 2, '04b')} | {format(int(valid_promo_2005) << 3, '04b')}")
print(f"Kode Biner   : {format(kode_transaksi_2005, '04b')}")
print(f"Kode Desimal : {kode_transaksi_2005}")

print("\n=== Pemeriksaan Status ===")

print("\nCek Member")
print(f"{format(kode_transaksi_2005, '04b')} & {format(int(valid_status_2005) << 0, '04b')}")
print(f"Hasil Biner  : {format((kode_transaksi_2005) & int(valid_status_2005) << 0, '04b')}")
print(f"Hasil Desimal: {(kode_transaksi_2005) & int(valid_status_2005) << 0}")

print("\nCek Promo")
print(f"{format(kode_transaksi_2005, '04b')} & {format(int(valid_promo_2005) << 3, '04b')}")
print(f"Hasil Biner  : {format((kode_transaksi_2005) & int(valid_promo_2005) << 3, '04b')}")
print(f"Hasil Desimal: {(kode_transaksi_2005) & int(valid_promo_2005) << 3}")

print("\n=== Perbandingan Status ===")
print(f"Kode Transaksi: {format(kode_transaksi_2005, '04b')}")
print(f"Kode Referensi: {format(kode_referensi_2005, '04b')}")
print(f"{format(kode_transaksi_2005, '04b')} ^ {format(kode_referensi_2005, '04b')}")
print(f"Hasil Biner  : {format((kode_transaksi_2005) ^ (kode_referensi_2005), '04b')}")
print(f"Hasil Desimal: {(kode_transaksi_2005) ^ (kode_referensi_2005)}")

print("\n=== Shift ===")
print(f"{format(kode_transaksi_2005, '04b')} << 1")
print(f"Hasil Biner  : {format((kode_transaksi_2005) << 1, '04b')}")
print(f"Hasil Desimal: {(kode_transaksi_2005) << 1}")

print("=== SELESAI ===")
```
