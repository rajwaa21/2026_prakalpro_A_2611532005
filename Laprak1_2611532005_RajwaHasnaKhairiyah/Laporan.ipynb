## 2.1 Dasar Teori
 Dasar teori ini akan membahas tiga komponen utama yang digunakan dalam praktikum, yaitu Python sebagai bahasa pemrograman, VS Code sebagai teks editor untuk menyusun kode, serta GitHub sebagai platform untuk menyimpan dan mengelola repositori proyek.

### 2.1.1. Python 
Python adalah bahasa pemrograman yang banyak digunakan dalam aplikasi web, pengembangan perangkat lunak, ilmu data, dan machine learning yang menggunakan interpreter untuk menjalankan kode programnya. Interpreter tersebut dapat menerjemahkan kode secara langsung, dan dapat dijalankan di berbagai platform. Python merupakan bahasa pemrograman tingkat tinggi (high-level programming) yang mudah dipahami dan digunakan karena program Phyton memiliki sintaksis dasar seperti dalam bahasa Inggris. Python diciptakan oleh Guido van Rossum pada 1991.

### 2.1.2 Visual Studio Code
Visual Studio Code merupakan aplikasi code editor untuk membantu proses pengembangan sebuah aplikasi yang dapat dijalankan pada perangkat desktop dengan basis Linux, Windows, dan Mac OS yang dikembangkan oleh Microsoft. VS Code adalah editor kode yang mendukung berbagai bahasa pemrograman, termasuk JavaScript, Python, C++, HTML, CSS, dll.

### 2.1.3. GitHub
GitHub adalah platform berbasis cloud yang memungkinkan tim untuk berbagi kode, meninjau perubahan, membahas masalah, dan mengelola proyek di satu tempat. GitHub merupakan layanan hosting berbasis web yang memfasilitasi pengembangan proyek software menggunakan sistem kontrol versi git. GitHub dapat mempermudah dalam melakukan kolborasi pengerjaan proyek, mencegah perubahan kode yang dapat merusak kode asli dan dapat digunakan sebagai portofolio bagi developer.

### 2.2.2 Membuat Repository GitHub
Setelah program selesai dibuat melalui VS Code, proyek dapat diintegrasikan dengan GitHub. Repository GitHub berfungsi sebagai tempat penyimpanan kode program.

Tahapan yang dilakukan meliputi:

1. Buat repository baru di GitHub.
2. Buka folder proyek melalui VS Code.
3. Buka terminal yang tersedia di VS Code.
4. Inisialisasi Git pada folder proyek.
5. Masukkan file ke dalam staging area.
6. Buat commit untuk menyimpan perubahan.
7. Hubungkan repository lokal dengan repository yang ada di GitHub.
8. Lakukan push untuk mengunggah file dan kode program ke repository GitHub.

Perintah yang digunakan:


```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

**Penjelasan :**
1. git init berfungsi untuk membuat repository Git baru di dalam folder proyek.
2. git add . berfungsi untuk memasukkan seluruh perubahan yang terdapat pada proyek ke dalam staging area.
3. git commit berfungsi untuk menyimpan perubahan yang telah dilakukan ke dalam repository disertai dengan keterangan.
4. git branch -M main berfungsi untuk menetapkan nama branch utama menjadi nama main.
5. git remote add origin digunakan untuk menghubungkan repository Git lokal dengan repository yang berada di GitHub.
6. git push berfungsi untuk mengirim atau mengunggah perubahan dari repository lokal ke repository GitHub.
