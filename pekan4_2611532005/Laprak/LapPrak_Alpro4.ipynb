## 2.1 Dasar Teori
Pada umumnya, sebuah program akan membaca dan menjalankan kode secara berurutan dari atas ke bawah. Namun, dalam pembuatan program yang sebenarnya, kita sering membutuhkan program yang lebih fleksibel yaitu program yang bisa mengambil keputusan sendiri tergantung pada situasi atau input yang dimasukkan oleh pengguna. Kemampuan program untuk memilih jalur kode mana yang harus dijalankan inilah yang dikenal dengan istilah Control Flow. Pada bagian ini, akan dibahas konsep dasar percabangan dalam bahasa pemrograman Python.

### 2.1.1 Conditional Statement
Pernyataan kondisional digunakan untuk mengontrol alur eksekusi dalam sebuah program berdasarkan kondisi tertentu. Pernyataan ini memungkinkan program untuk mengeksekusi blok kode yang berbeda tergantung pada apakah suatu kondisi bernilai Benar atau Salah.

### 2.1.2 Statement if
Pernyataan if mengevaluasi suatu kondisi (ekspresi yang menghasilkan True atau False ). Jika kondisinya benar, blok kode di dalam pernyataan if akan dieksekusi. Jika kondisinya salah, blok kode akan dilewati. Ini berguna ketika hanya satu pernyataan yang perlu dieksekusi.

### 2.1.3 Statement if-else
Pernyataan If Else digunakan untuk mengeksekusi satu blok kode ketika kondisinya Benar dan blok kode lain ketika kondisinya Salah. Dalam pemrograman, struktur kendali percabangan digunakan untuk mengatur jalannya kode sesuai dengan kondisi logika tertentu.

### 2.1.4 Statement if-elif-else
Pernyataan if-elif-else adalah struktur kendali percabangan bertingkat yang digunakan ketika program harus mengevaluasi lebih dari dua kemungkinan kondisi logika. Pernyataan elif digunakan untuk memeriksa beberapa kondisi dalam sebuah program. Pernyataan ini mengeksekusi blok kode ketika kondisinya bernilai True setelah kondisi sebelumnya bernilai False. 

### 2.1.5 Match Case
Match case di Python digunakan untuk melakukan pencocokan pola. Fitur ini membandingkan suatu nilai dengan beberapa pola yang tersedia, lalu mengeksekusi blok kode yang sesuai dengan kecocokan pertama. Dengan adanya match-case, Anda memiliki alternatif yang lebih rapi dan terorganisir dibandingkan menggunakan struktur pernyataan if-elif-else yang panjang saat menangani banyak kondisi.

### 2.1.6 Multi if
Multi If adalah sekumpulan pernyataan if yang ditulis secara terpisah dan dievaluasi satu per satu dari atas ke bawah. Pada multi if setiap kondisi bersifat independen dan tidak saling membatalkan. Jika ada beberapa kondisi yang bernilai True, semuanya akan dijalankan.

### 2.1.7 Nested if
Pernyataan kondisional bersarang/Nested If adalah struktur percabangan di mana terdapat pernyataan if di dalam blok if yang lain. Program akan mengevaluasi kondisi terluar terlebih dahulu. Jika kondisi terluar bernilai True, barulah kondisi di bagian dalamnya dievaluasi.

### 2.1.8 Operator Relasional & Operator Logika
Operator relasional adalah operator yang berfungsi untuk membandingkan dua buah nilai, variabel, atau operan. Tujuan utama dari penggunaan operator ini adalah untuk menentukan hubungan atau relasi antara kedua entitas yang dibandingkan, seperti apakah nilainya sama, lebih besar, atau lebih kecil. Bahasa pemrograman Python menyediakan enam jenis operator relasional dasar, yaitu: Sama dengan (==), Tidak sama dengan (!=), Lebih besar dari (>), Lebih kecil dari (<), Lebih besar atau sama dengan (>=), Lebih kecil atau sama dengan (<=), 

Operator logika digunakan untuk menggabungkan pernyataan kondisional. Python memiliki tiga operator logika yaitu, and untuk engembalikan True jika kedua pernyataan tersebut benar, or untuk mengembalikan True jika salah satu pernyataan benar dan not untuk nembalikkan hasilnya, mengembalikan False jika hasilnya benar

## 2.2 Kode Program

### 2.2.1 Program "if1_2611532005"

```python
# Buat file dengan nama if1_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

ipk_2005 = float(input("Input IPK Anda = "))

if ipk_2005 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str (ipk_2005))

print("Program Selesai")
```

**Penjelasan :**

2. Nilai IPK dibandingkan dengan angka 2.75 menggunakan percabangan tunggal if.
3. Jika nilai IPK lebih besar dari 2.75, maka menghasilkan nilai True dan program menampilkan pesan "Anda Lulus Sangat Memuaskan dengan IPK".
4. Nilai IPK pada pesan kelulusan diubah menjadi tipe string menggunakan str() agar dapat digabungkan dengan teks kalimat sebelumnya.
5. Jika nilai IPK kurang dari atau sama dengan 2.75, maka menghasilkan nilai False dan program mengabaikan blok percabangan tersebut.
6. Program menampilkan teks "Program Selesai" sebagai penanda berhentinya eksekusi, yang akan selalu dicetak terlepas dari terpenuhi atau tidaknya kondisi pada percabangan if.

### 2.2.5 Program "multi_if2_2611532005"

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_2005 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2005 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member_2005 = input_member_2005 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2005 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2005 = input_promo_2005 in ["y", "ya"]

total_diskon_persen_2005 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2005 > 1000000:
    total_diskon_persen_2005 += 10 # Diskon belanja besar

if is_member_2005:
    total_diskon_persen_2005 += 5 # Diskon member

if kode_promo_valid_2005:
    total_diskon_persen_2005 += 15  # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_2005 = total_belanja_2005 * (total_diskon_persen_2005 / 100)
total_bayar_2005 = total_belanja_2005 - nominal_diskon_2005

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon  : {total_diskon_persen_2005}% (Rp {nominal_diskon_2005:,.0f})")
print(f"Total Bayar   : Rp {total_bayar_2005:,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_2005}%")
# Output: Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan :**
1. Memasukkan data total belanja. Data yang diinputkan pengguna diubah menjadi tipe bilangan desimal menggunakan float() dan disimpan pada variabel total_belanja_2005.
2. Meminta input status keanggotaan (member). Teks yang dimasukkan diproses menggunakan fungsi .strip() untuk menghilangkan spasi tidak sengaja di awal/akhir, dan .lower() untuk menyeragamkan huruf menjadi kecil semua. Program mengevaluasi apakah teks tersebut ada di dalam daftar ["y", "ya"]. Hasil evaluasi berjenis boolean disimpan pada variabel is_member_2005.
3. Meminta input validitas kode promo dengan menerapkan boolean yang sama seperti pada status member, lalu menyimpannya pada variabel kode_promo_valid_2005.
4. Mendeklarasikan variabel total_diskon_persen_2005 dengan nilai awal 0 yang berfungsi sebagai akumulator persentase diskon.
5. Program menggunakan Multi-If (If beruntun yang independen/terpisah). Struktur Multi-If ini memeriksa setiap kondisi dari atas ke bawah tanpa saling membatalkan.
6. Pada evaluasi percabangan pertama, jika belanja lebih dari 1.000.000, nilai diskon bertambah 10. Pada percabangan kedua, jika pengguna adalah member, diskon bertambah 5. Pada percabangan ketiga, jika memiliki kode promo valid, diskon bertambah 15. Jika semua syarat terpenuhi, seluruh diskon akan terakumulasi di dalam variabel menjadi 30%.
7. Menghitung nilai nominal_diskon_2005 dengan mengalikan total belanja dengan persentase diskon yang telah dibagi 100. Setelah itu, menghitung total_bayar_2005 dengan cara mengurangkan total belanja dengan nominal diskon.
8. Menampilkan rincian pembayaran ke layar menggunakan pemformatan f-string. Sintaks :,.0f di dalam kurung kurawal digunakan untuk mengonversi angka menjadi format mata uang yang rapi, di mana , bertindak sebagai pemisah ribuan dan .0f memastikan tidak ada angka desimal berlebih di belakang koma.

### 2.2.6 Program "nilai_2611532005"

```python
# Buat file dengan nama nilai_nim.py
# Buat program untuk nested if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi nilai angka menjadi huruf

nilai_2005 = int(input("Inputkan nilai angka= "))

if nilai_2005 >= 81:
    print("A")
elif nilai_2005 >= 70:
    print("B")
elif nilai_2005 >= 60:
    print("C")
elif nilai_2005 >= 50:
    print("D")
else:
    print("E")
```

**Penjelasan :**

1. Meminta pengguna memasukkan nilai berupa angka. Data teks yang diinputkan dikonversi menjadi tipe bilangan bulat menggunakan int() dan disimpan pada variabel nilai_2005.
2. Program mengevaluasi nilai tersebut menggunakan struktur percabangan banyak kondisi (if-elif-else) untuk menentukan konversi rentang nilai angka menjadi huruf. Struktur ini bekerja secara berurutan dari atas ke bawah.
3. Pada pengecekan pertama (if), program mengevaluasi apakah nilai lebih besar dari atau sama dengan 81. Jika True, program mencetak "A" dan seluruh evaluasi di bawahnya akan langsung dilewati.
4. Jika kondisi pertama bernilai False, program turun ke pengecekan kedua (elif). Kondisi ini hanya memeriksa apakah nilai lebih dari atau sama dengan 70. Jika True, program mencetak "B".
5. Jika kondisi kedua bernilai False, program turun ke pengecekan ketiga (elif). Jika nilai lebih dari atau sama dengan 60, program mencetak "C".
6. Jika kondisi ketiga bernilai False, program turun ke pengecekan keempat (elif). Jika nilai lebih dari atau sama dengan 50, program mencetak "D".
7. Jika semua kondisi if dan elif di atas bernilai False, program akan menjadikan blok else sebagai jalan keluar terakhir dan mencetak "E".

### 2.2.7 Program "match_case1_2611532005"

```python
# Buat file dengan nama match_case1_nim.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulam

bulan_2005 = int(input("Masukkan angka bulan (1-12): "))

match bulan_2005:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Bulan tidak valid, masukkan angka 1-12.")
```

**Penjelasan :**

1. Meminta pengguna memasukkan angka representasi bulan. Data teks yang diinputkan dikonversi menjadi tipe bilangan bulat menggunakan fungsi int() dan disimpan pada variabel bulan_2005.
2. Program mengevaluasi nilai dari variabel tersebut menggunakan struktur match-case.
3. Program akan mencocokkan isi variabel bulan_2005 dengan setiap blok case yang tersedia. Jika nilainya persis sama dengan angka pada salah satu case (misalnya 8 untuk mencetak "Agustus"), program akan mengeksekusi blok tersebut lalu secara otomatis menghentikan proses pencarian ke bawah.
4. Jika angka yang diinputkan pengguna di luar rentang 1 hingga 12, maka tidak ada satupun case spesifik yang terpenuhi.
5. Program menggunakan case _ (berfungsi sama seperti else sebagai penangkap nilai default atau wildcard). Di dalam blok ini, program menampilkan pesan "Bulan tidak valid, masukkan angka 1-12." sebagai bentuk penanganan kesalahan.

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

nama_2005 = input("Masukkan Nama Pengunjung : ")
umur_2005 = int(input("Input umur anda : "))
sim_2005 = input("Apakah Anda Sudah Punya SIM C (y/t) : ").strip().lower()

# Pilihan paket wahana:
# case 1: "Wahana Safari Rimba" | Harga Satuan: Rp 50.000
# case 2: "Wahana Arung Jeram" | Harga Satuan: Rp 75.000
# case 3: "Wahana Motor ATV Ekstrim" | Harga Satuan: Rp 120.000
# case 4: "Wahana Roller Coaster Kilat" | Harga Satuan: Rp 100.000
# case 5: "Wahana All-Access VIP" | Harga Satuan: Rp 220.000
# case _: Cetak pesan "Paket wahana tidak valid!" dan hentikan eksekusi transaksi.

print("\nPilihan Paket Wahana (1-5):")
print("1. Safari Rimba          (Rp 50,000)")
print("2. Arung Jeram           (Rp 75,000)")
print("3. Motor ATV Ekstrim     (Rp 120,000)")
print("4. Roller Coaster Kilat  (Rp 100,000)")
print("5. All-Access VIP        (Rp 220,000)")

paket_2005 = int(input("Masukkan nomor paket (1-5) : "))

match paket_2005:
    case 1:
        wahana_2005 = "Safari Rimba"
        harga_satuan_2005 = 50000
    case 2:
        wahana_2005 = "Arung Jeram"
        harga_satuan_2005 = 75000
    case 3:
        wahana_2005 = "Motor ATV Ekstrim"
        harga_satuan_2005 = 120000
    case 4:
        wahana_2005 = "Roller Coaster Kilat"
        harga_satuan_2005 = 100000
    case 5:
        wahana_2005 = "All-Access VIP"
        harga_satuan_2005 = 220000
    case _:
        print("Paket wahana tidak valid!")
        exit()

jumlah_tiket_2005 = int(input("Masukkan jumlah tiket : "))

member_2005 = input("Apakah Anda member? (y/t) : ").strip().lower()
kode_promo_2005 = input("Apakah kode promo valid? (y/t) : ").strip().lower()

print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

# Khusus untuk Paket 3 (Motor ATV Ekstrim), evaluasi izin pengendara:
# if umur >= 17 and sim == 'y', Cetak: "Anda sudah dewasa dan boleh mengendarai ATV sendiri."
# elif umur >= 17 and sim != 'y', Cetak: "Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur)."
# elif umur < 17 and sim == 'y, Cetak: "Identitas tidak valid: Belum cukup umur memiliki SIM."
# else:, Cetak: "Anda belum cukup umur dan tidak boleh bawa motor ATV."

if paket_2005 == 3:
    if umur_2005 >= 17 and sim_2005 == 'y':
        print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
    elif umur_2005 >= 17 and sim_2005 != 'y':
        print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
    elif umur_2005 < 17 and sim_2005 == 'y':
        print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
    else:
        print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")

# Untuk paket selain 3, gunakan if-else sederhana untuk mengecek apakah umur >= 10 tahun).
else:
    if umur_2005 >= 10:
            print("Status Akses: Anda memenuhi syarat umur untuk", wahana_2005)
    else:
            print("Status Akses: Anda belum memenuhi syarat umur untuk", wahana_2005)

# Hitung subtotal = harga_satuan * jumlah_tiket.
# if subtotal >= 200000: total_diskon_persen += 10 (Diskon Belanja Besar)
# if is_member in ['y', 'ya']: total_diskon_persen += 5 (Diskon Member)
# if kode_promo_valid in ['y', 'ya']: total_diskon_persen += 15 (Diskon Voucher Promo)
# if jumlah_tiket >= 5: total_diskon_persen += 5 (Diskon Tambahan Rombongan)

subtotal_2005 = harga_satuan_2005 * jumlah_tiket_2005
diskon_2005 = 0

if subtotal_2005 >= 200000: 
    diskon_2005 += 10 # Diskon Belanja Besar
if member_2005 in ['y', 'ya']:
    diskon_2005 += 5 # Diskon Member
if kode_promo_2005 in ['y', 'ya']:
    diskon_2005 += 15 # Diskon Voucher Promo
if jumlah_tiket_2005 >= 5:
    diskon_2005 += 5 # Diskon Tambahan Rombongan

nominal_diskon_2005 = subtotal_2005 * (diskon_2005 / 100)
total_bayar_2005 = subtotal_2005 - nominal_diskon_2005

print("\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {subtotal_2005:,.0f}")
print(f"Total Diskon     : {diskon_2005}% (Rp {nominal_diskon_2005:,.0f})")
print(f"Total Bayar      : Rp {total_bayar_2005:,.0f}")

if total_bayar_2005 > 300000:
    print("Catatan Layanan  : Selamat! Anda berhak mendapatkan Souvenir Gratis.")
else:
    print("Catatan Layanan  : Terima kasih telah berkunjung.")

print("Program Selesai")
```

<h1 align="center" > BAB III <br> KESIMPULAN </h1>

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa penggunaan percabangan (control flow) membuat jalan sebuah program menjadi lebih dinamis, tidak sekadar lurus berurutan dari atas ke bawah. Program kini bisa merespons dan memberikan hasil yang berbeda-beda tergantung pada data yang dimasukkan oleh pengguna. Untuk membuat menu pilihan yang angkanya sudah pasti penggunaan match-case terbukti lebih rapi dan efisien daripada menulis if-elif. Sementara itu, untuk mengecek aturan yang berlapis kita bisa menggunakan nested if. Dengan cara ini, program hanya akan mengecek syarat lanjutan jika syarat pertamanya sudah terpenuhi.

Selain itu, praktikum ini juga menunjukkan bahwa kita bisa melakukan perhitungan yang ditumpuk dengan menggunakan multi-if. Tujuannya agar program mengecek semua syarat satu per satu tanpa ada syarat yang terlewat atau saling membatalkan. Agar program tidak mudah error saat pengguna salah ketik (huruf besar/kecil atau ada spasi lebih), kita bisa mengatasinya dengan fungsi .strip() dan .lower(). Hasil akhir perhitungannya pun menjadi lebih rapi seperti format uang asli berkat bantuan f-string. Secara keseluruhan, pemahaman tentang logika percabangan ini sangat penting sebagai dasar untuk membuat program otomatis yang fleksibel dan bisa diterapkan langsung di dunia nyata.
